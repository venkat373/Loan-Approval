# 1.Buisiness Understanding

### Problem Statement

A bank wants to predict whether a loan application should be approved based on the applicant's personal and financial information. This helps reduce manual effort and speed up the loan approval process

### Buisiness Objective

* Reduce manual verification
* Minimize the risk of Approval high - risk applicants
* Improve Customers experience

### Machine Learning Problem

* Type: Supervised Learning
* Category: Binary Classification
* Target Variable: Loan_Status

# 2.Dataset Understanding

--->These are the main Features/Columns available in the dataset :

* No_of_dependents – Number of dependents the applicant has

* Education – Graduate / Not Graduate

* Self_employed – Whether the applicant is self-employed

* Income_annum – Annual income of the applicant

* Loan_amount – Loan amount requested

* Loan_term – Loan repayment time (years)

* Cibil_score – Credit score indicating creditworthiness

* Residential_assets_value – Value of residential assets

* Commercial_assets_value – Value of commercial assets

* Luxury_assets_value – Value of luxury assets

* Bank_asset_value – Bank balance / financial assets

* Loan_status (Target variable) – Approved (1) or Rejected (0)

Target_variable:
* Approved
* Rejected

# 8.Data Spliting

```python
from sklearn.model_selection import train_test_split
x=df.drop(columns=['loan_status'])
y=df['loan_status']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

# 9.Model Bilding

-->> Train multiple classification algorithms compare their performances and identify the best model for loan approval

```python
from sklearn.metrics import accuracy_score
```

### A.Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train,y_train)
```

```python
# Prediction
y_pred_lr=lr.predict(x_test)
```

```python
# Accuracy
accuracy_score(y_test,y_pred_lr)
```

### B.Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
DT=DecisionTreeClassifier(random_state=42)
DT.fit(x_train,y_train)
```

```python
# Prediction
y_pred_DT=DT.predict(x_test)
# Accuracy
accuracy_score(y_test,y_pred_DT)
```

### C.Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
RF=RandomForestClassifier(n_estimators=100,max_depth=15,min_samples_split=2)
RF.fit(x_train,y_train)
```

```python
# Prediction
y_pred_RF=RF.predict(x_test)
# Accuracy
accuracy_score(y_test,y_pred_RF)
```

### D.K-Nearest Neighbors(KNN)

```python
from sklearn.neighbors import KNeighborsClassifier
KNN=KNeighborsClassifier(n_neighbors=5)
KNN.fit(x_train,y_train)
```

```python
# Prediction
y_pred_KNN=KNN.predict(x_test)
# Accuracy
accuracy_score(y_test,y_pred_KNN)
```

### E.Support Vector Machine(SVM)

```python
from sklearn.svm import SVC
SVC=SVC(kernel='rbf')
SVC.fit(x_train,y_train)
```

```python
# Prediction
y_pred_SVC=SVC.predict(x_test)
# Accuracy
accuracy_score(y_test,y_pred_SVC)
```

### F.Ada Boost

```python
from sklearn.ensemble import AdaBoostClassifier
ada=AdaBoostClassifier(n_estimators=100,random_state=42)
ada.fit(x_train,y_train)
```

```python
# Prediction
y_pred_ada=ada.predict(x_test)
# Accuracy
accuracy_score(y_test,y_pred_ada)
```

### G.XGBoost

```python
from xgboost import XGBClassifier
xgb=XGBClassifier(
    random_state=42,
    eval_metrics='logloss'
)
xgb.fit(x_train,y_train)
```

```python
# Prediction
y_pred_xgb=xgb.predict(x_test)
# Accuracy
accuracy_score(y_test,y_pred_xgb)
```

# Compare All Models

```python
results = pd.DataFrame({
    'Models': [
        'Logistic Regression', 
        'Decision Tree', 
        'Random Forest', 
        'KNN', 
        'SVM', 
        'AdaBoost', 
        'XGBoost'
    ], 
    'Accuracy': [
        accuracy_score(y_test, y_pred_lr), 
        accuracy_score(y_test, y_pred_DT), 
        accuracy_score(y_test, y_pred_RF), 
        accuracy_score(y_test, y_pred_KNN), 
        accuracy_score(y_test, y_pred_SVC),
        accuracy_score(y_test, y_pred_ada), 
        accuracy_score(y_test, y_pred_xgb)
    ]
})

results = results.sort_values(by='Accuracy', ascending=False)
results
```

# 10.Hyperparameter Tuning

-->Hyperparameter tuning is the process of finding best parameter values for a ML Model

to improve its performance

* Improve model accuracy
* Reduce overfitting.
* Find the best model automatically

#### method1: Grid search cv

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
```

```python
# Parameters Grid
param_grid={
    'n_estimators':[50,100,200],
    'max_depth':[3,10,15],
    'min_samples_split':[2,5,10]
}
```

```python
# Apply Grid
rf=RandomForestClassifier(random_state=42)

grid=GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)
grid.fit(x_train,y_train)
```

```python
# Best parameters
print('Best Parameters:',grid.best_params_)
```

```python
print('Best Score:',grid.best_score_)
```

```python
# Best Model
best_rf=grid.best_estimator_
y_pred=best_rf.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print('Test Accuracy:',accuracy_score(y_test,y_pred))
```

#### Method2:RandomizedSearchCV

```python
from sklearn.model_selection import RandomizedSearchCV
param_dist = {
    'n_estimators': [50,100,150,200,250,300],
    'max_depth': [5,10,15,20,None],
    'min_samples_split':[2,5,10],
    'min_samples_leaf':[1,2,4]
}
```

```python
# Apply Model
```

```python
random_search=RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_dist,
    n_iter=10,
    cv=5,
    scoring='accuracy',
    random_state=42
)
```

```python
# Best Parameters
random_search.fit(x_train,y_train)
```

```python
# Best Model
best_model=random_search.best_estimator_
y_pred=best_model.predict(x_test)
```

```python
# Best parameters
print('Best Parameters:',random_search.best_params_)
```

```python
print('Test Accuracy:',accuracy_score(y_test,y_pred))
```
